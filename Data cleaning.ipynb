# HOSPITAL MANAGEMENT DATA CLEANING

```python
# Import libraries
 
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
# Load dataset


df = pd.read_csv(r"d:\my multiple projects\end to end hospital managment project\hospital_management_dataset.csv")
```

```python
# Verify that load the dataset

df.head()
```

```python
# Quikly understand the dataset

df.info()
```

```python
# Apply descriptive statistics

df.describe()
```

```python
# Standarize column names

df.columns = df.columns.str.strip().str.replace(" ","_").str.lower()

# Verify that standarize the column names

df.columns
```

```python
# Check data types

df.dtypes
```

STANDARIZE THE COLUMN DATA TYPES

```python
# Convert all date columns to proper datetime format

date_col = ["appointment_date","admission_date","discharge_date","payment_date"]

for col in date_col:
    
    df[col] = pd.to_datetime(df[col],errors="coerce")

# Convert all time columns to proper datetime format

time_col = ["appointment_time","check_in_time","consultation_start_time","consultation_end_time"]

for col in time_col:
    
    df[col] = pd.to_datetime(df[col],format = "%H:%M:%S",errors="coerce")


```

```python
# Verify the conversion of the date and time columns

df.dtypes

df.head()
```

```python
# Convert all float columns to proper float format

float_col = ["doctor_fee","medicine_cost","test_cost","room_charges","total_bill"]

for col in float_col:
    
    df[col] = df[col].astype(float)
```

```python
# Validate the datatypes

df.dtypes
```

Handle missing values

```python
# Detect missing values

df.isnull().sum()
```

Fill missing values

```python
# Fill missing phone no

df["phone_no"] = df["phone_no"].fillna("Not Available")

# Fill missing cities

df["city"] = df["city"].fillna("Unknown")

# Fill missing values in medicine column

df["medicine"] = df["medicine"].fillna("No Medicine")

# Fill missing values in the lab test column

df["lab_test"] = df["lab_test"].fillna("No Test")
```

```python

df["payment_status"].unique()

# Replace missing payment methods for paid patients

df.loc[(df["payment_status"] == 'Paid') & 
        
        (df["payment_method"].isna()), 
        
        "payment_method"] = "UnKnown"

# Replace missing payment methods for Unpaid patients

df.loc[(df["payment_status"] == 'Unpaid') & 
        
        (df["payment_method"].isna()), 
        
        "payment_method"] = "Not Paid"
```

```python
# Validate missing values after cleaning

df.isnull().sum()
```

Handle duplicate records

```python
# Detect duplicate records

df.duplicated().sum()
```

```python
# Drop duplicate records

df = df.drop_duplicates()
```

```python
# Validate duplicate records after droping

df.duplicated().sum()
```

```python
df.head(1)
```

Create function for cleaning categorical columns

```python
# Create function to clean whitespace and standardize text case

def clean_column(col):
    
    df[col] = df[col].str.strip().str.title()
    
```

```python
# Clean appointment status column

df["appointment_status"].unique()

clean_column("appointment_status")
```

```python
# Clean gender column

df["gender"].unique()

clean_column("gender")
```

```python
# Clean payment status column

df["payment_status"].unique()

clean_column("payment_status")
```

```python
# Clean payment method column

df["payment_method"].unique()

clean_column("payment_method")
```

Standardize the dataset and remove negative values in bill and age columns

```python
# Remove negetive age and age greater that one hundred twenty

df = df[(df["age"] > 0) & (df["age"] < 120)]
```

```python
# Remove negetive billing records from the data set

df = df[df["total_bill"] > 0]
```
